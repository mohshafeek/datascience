```python
## f-string
```

```python
first_name = "fName"
last_name = "lName"

print(f"{first_name} {last_name}")
```

```python
price = 19.99999999
print(f"Price: {price:.2f}") # .2f = 2 decimals
```

# Booleans and comparisons

```python
x = 10
y = 20
```

```python
x == y # Equals Op
```

```python
x != y # Not equals
```

```python
print(x > y)
print(x < y)
print(x >= y)
print(y <= x)
```

```python
age = 17
is_audit = age >= 18
has_id = True

# Logical Op - and, or, not

print("Can enter:", is_audit and has_id)
print("Not a weekend:", not False)
```

```python
print(False and True)
print(False and False)
print(True and True)
print(True or False)
print(False or False)
print(not True)
print(not False)
```
