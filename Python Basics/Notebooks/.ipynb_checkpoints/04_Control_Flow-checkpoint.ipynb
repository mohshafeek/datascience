# Control Flow

Assignment: modifying student dataset and checking stats using loops/ifs.

```python
# initial marks list
students = {
    "student1": 45,
    "student2": 72,
    "student3": 85,
    "student4": 55,
    "student5": 68
}

# add 5 more students
students["student6"] = 90
students["student7"] = 38
students["student8"] = 82
students["student9"] = 61
students["student10"] = 88

print("all students:", students)
```

```python
# change pass mark from 50 to 60
pass_mark = 60
print("new passing mark:", pass_mark)
```

```python
# count how many got atleast 80
count = 0
for name in students:
    if students[name] >= 80:
        count = count + 1

print("students with 80+ marks:", count)
```

```python
# Display the highest mark
highest = 0
for name in students:
    if students[name] > highest:
        highest = students[name]

print("Highest mark in class:", highest)
```

```python
# Display percentage of students passed
passed_students = 0
total = len(students)

for name in students:
    if students[name] >= pass_mark:
        passed_students = passed_students + 1

percentage = (passed_students / total) * 100
print("Percentage of students who passed:", percentage, "%")
```
