### assignment 1 - word & char counter

```python
sentence = input("Enter a sentence: ")

total_chars = len(sentence)
print("Total chars in sentence:", total_chars)

# split into list
words = sentence.split()
print("Total number of words:", len(words))

print("--------")
# loop through each word and print its length
for w in words:
    print(w, "->", len(w), "chars")
```

### assignment 2 - Todo-list App

```python
my_tasks = []

while True:
    task = input("enter new task: ")
    my_tasks.append(task)
    
    # ask user for confirmation to stop
    check = input("do u have more tasks? (yes/no): ")
    
    if check.lower() == 'no' or check.lower() == 'n':
        break

print("\nMy Tasks :")
num = 1
for item in my_tasks:
    print(num, ".", item)
    num = num + 1
```
